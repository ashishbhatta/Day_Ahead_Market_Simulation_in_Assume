Data from January 1 2026 to January 31 2026 is used for creating a demad file for simulation.

The data is extrated using the script https://github.com/ashishbhatta/Day_Ahead_Market_Simulation_in_Assume/tree/main/IEX_scraping

The data is stored in https://github.com/ashishbhatta/Day_Ahead_Market_Simulation_in_Assume/blob/main/IEX_Aggrigated_suply_demand.csv


```python
import pandas as pd
import numpy as np
```

```python

df = pd.read_csv("https://raw.githubusercontent.com/ashishbhatta/Day_Ahead_Market_Simulation_in_Assume/refs/heads/main/IEX_Aggrigated_suply_demand.csv")
df.tail()
```

```python
df.head()
```

```python
# In the Date column some date are in format d/m/y while others are in d-m-y so converting them to the same format.

df['Date'] = df['Date'].str.replace('-', '/')
```

```python
# Creating a new column "datetime" containg the date and time information.

df[['Start_Time', 'End_Time']] = df['Time Block'].str.split('-', expand=True)
df['datetime'] = pd.to_datetime(
    df['Date'] + ' ' + df['Start_Time'], format='%d/%m/%Y %H:%M'
)
df.drop(['Source URL','Time Block','Date'],axis=1,inplace=True)
col = df.pop('datetime')
df.insert(0, 'datetime', col)
df.tail()
```

```python
df.tail()
```

```python
df2 = df.pivot_table(index = "datetime", columns = "Price (in Rs./MWh)",values="Buy / Demand (in MW)")
df2.columns.name = None # used to remove multi index column
```

```python
df2.tail()
```

```python
# Subtracting the demand from adjascent demand coulumn

# 1. Identify price columns in order (excluding datetime)
price_columns = [col for col in df2.columns if col.lower() != 'datetime']

# 2. Subtract adjacent columns horizontally
for i in range(len(price_columns)):
    col1 = price_columns[i]

    if i < len(price_columns) - 1:
        col2 = price_columns[i + 1]
        # Column_i - Column_{i+1}
        df2[f'Demand_{i+1}'] = (df2[col1] - df2[col2]).abs().round(2)
    else:
        # Last column: Column_10 - 0 (value remains unchanged)
        df2[f'Demand_{i+1}'] = df2[col1].abs().round(2)

df2.drop(columns=df2.columns[:10], inplace=True)
df2.head()
```

```python
df2.tail()
```

```python
df2.to_csv('demand_df.csv')
```
